## What the bound means
Every edge costs at least walking time. An outside candidate requires reaching the expanded outer boundary first. A 0.99 projection safety factor gives a conservative straight-line lower bound. If a chosen effort budget is below that bound, an outside candidate or detour cannot improve a qualifying result under this model. This does not verify all real paths or entrances.

```python
b=pd.read_csv(ACCESS/'block_accessibility.csv',dtype={'GEOID20':str})
housing=(b.HU100>0)&(b.access_snap_gap_m<=100)
assert b.loc[housing,'boundary_exit_lower_bound_min'].gt(30).all()
print('Minimum eligible exit lower bound:',b.loc[housing,'boundary_exit_lower_bound_min'].min())
display(Image(filename=str(ROOT/'output/plano_city/public/plano_public_everyday_walks.png'),width=950))
```
